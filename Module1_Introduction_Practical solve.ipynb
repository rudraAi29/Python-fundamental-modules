### B.3 Type Conversion (Casting)
`int()`, `float()`, `str()` perform **explicit** conversion. Python performs
**implicit** conversion automatically when combining compatible types.

```python
a = "25"
b = 4.0

c = int(a) + int(b)     # explicit conversion: 25 + 4
print("int(a) + int(b) =", c, type(c))

d = a + str(b)             # explicit conversion the other way
print("a + str(b) =", d, type(d))

e = True + 1                  # implicit: bool -> int
print("True + 1 =", e, type(e))

# Uncommenting the next line raises a TypeError -- Python never silently
# mixes str with numeric types:
# f = a + b
```

## Part C: Operators & Precedence

### C.1 Arithmetic, Comparison, and Logical Operators

```python
a, b = 15, 4

print("Arithmetic:")
print(a + b, a - b, a * b, a / b)   # / always returns a float
print(a // b, "  <- floor division")
print(a % b, "  <- modulus (remainder)")
print(a ** 2, "  <- exponent")

print("\nComparison:")
print(a > b, a == b, a != b)

print("\nLogical:")
print((a > b) and (b > 0))
print((a > b) or (b > 100))
print(not (a > b))
```

### C.2 Assignment Shortcuts

```python
x = 10
x += 5    # same as x = x + 5
print(x)
x -= 3
print(x)
x *= 2
print(x)
x //= 4
print(x)
```

### C.3 Operator Precedence
Order (high to low): `()` &gt; `**` &gt; `*, /, //, %` &gt; `+, -` &gt; comparisons &gt; `not` &gt; `and` &gt; `or`

```python
result = 5 + 3 * 2 ** 2 - (4 / 2) # left to right when we will have to do operation of same class(+,- or*,/)
print("result =", result)
print("type   =", type(result))   # float, because / always returns a float in Python 3
```

**Q6 (Medium).** Write a program that takes the radius of a circle as input and prints its area and circumference, correctly rounded to 2 decimal places (look up the `round()` built-in function).

```python
# Q6
import math

radius = float(input("Enter the radius of the circle: "))

area = math.pi * pow(radius,2)
circumference = 2 * math.pi * radius

print(f"Area: {round(area, 2)}")
print(f"Circumference: {round(circumference, 2)}")
```

**Q7 (Hard).** Write a program that takes a temperature in Celsius as input and converts it to both Fahrenheit and Kelvin, printing all three values in one formatted line using an f-string.

```python
# Q7
celsius = int(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15

print(f"{celsius} degree C is equivalent to {fahrenheit} degree F and {kelvin}k")
```

**Q8 (Hard).** A store applies a 12% tax and then a 5% discount (in that order) to an item's price. Write a program that takes the base price as input and prints the final price, correctly demonstrating operator precedence and order of operations in your calculation.

```python
# Q8
base_price = float(input("Enter base price: "))

# 12% tax means multiplying by 1.12. 
# 5% discount means retaining 95% of the price, so multiplying by 0.95.
final_price = base_price * 1.12 * 0.95

print(f"Final price after tax and discount: {round(final_price, 2)}")
```

## Part E: Mini-Challenge

**Simple Interest Calculator.** Write a program that:
1. Asks the user for Principal amount, annual interest Rate (%), and Time (in years)
2. Computes Simple Interest using $SI = \dfrac{P \times R \times T}{100}$
3. Computes the Total Amount ($= P + SI$)
4. Prints both values, formatted to 2 decimal places, in a neat summary

This combines everything from today: input, type conversion, arithmetic operators, and
formatted output.

```python
p = float(input("Enter Principal amount: "))
r = float(input("Enter annual interest Rate (%): "))
t = float(input("Enter Time (in years): "))

# Simple Interest Formula
si = (p * r * t) / 100
total_amount = p + si

print("\n--- Summary ---")
print(f"Simple Interest: {si:.2f}")
print(f"Total Amount: {total_amount:.2f}")
```
