### A.3 The `elif` Ladder

````python
score = 75
if score >= 90:
    grade = 'A'
elif score >= 75:
    grade = 'B'
elif score >= 60:
    grade = 'C'
else:
    grade = 'F'
print(grade)   # Output: B
````

### A.4 Worked Example: Grade Calculator
Trace the elif ladder above for `score = 75`. Which condition is checked first, and why do the later ones get skipped?

````python
score = 75
if score >= 90:
    print("Checking >= 90:", score >= 90)
elif score >= 75:
    print("Checking >= 75:", score >= 75, "-> this branch executes")
elif score >= 60:
    print("Checking >= 60: never reached, since a branch already matched")
else:
    print("else: never reached")
````

## Part B: Loops — For and While

### B.1 The `for` Loop and `range()`
`range(start, stop, step)` — `stop` is always **excluded**.

````python
for i in range(1, 6):     # 1, 2, 3, 4, 5 (stop excluded)
    print(i, end=" ")
print()

for ch in "Py":
    print(ch)

for i in range(10, 0, -2):   # counting down by 2
    print(i, end=" ")
````

### B.2 The `while` Loop

````python
count = 1
while count <= 5:
    print(count, end=" ")
    count += 1
print()
````

### B.3 Worked Example: Sum of Digits (while loop)
Trace this by hand for `num = 1234` before running.

````python
num = 1234
total = 0
while num > 0:
    digit = num % 10
    total += digit
    num = num // 10

print("Sum of digits:", total)   # Expect: 10
````

### B.4 Nested Loops: Pattern Printing

````python
for i in range(1, 5):
    for j in range(i):
        print("*",end=" ")
    print()
````

**Try it yourself:** Modify the pattern above to print an *inverted* triangle (4 stars on the first row, shrinking to 1).

````python
for i in range(4, 0, -1):
    for j in range(i):
        print("*",end=" ")
    print()
````

**in pyramid shape**

````python
row=4
for i in range(1,5): #print the spaces
    for j in range(row-i):
        print(" ",end="")
    for k in range (i): #print the stars
        print("*",end=" ")
    #moves to the next line
    print()
````

**in inverted pyramid**

````python
rows=4
for i in range(4,0,-1):
    for j in range (rows-i):
        print(" ",end="")
    for k in range (i):
        print("*",end=" ")
    print()
````

### B.5 Nested Loops: Multiplication Table

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i*j}", end="   ")
    print()
````

**Q4 (Medium).** Write a program using a `while` loop to check whether a given number is a prime number.

````python
# Q4
num = int(input("Enter a number: "))

if num <= 1:
    print("num is not a prime number")
else:
    i = 2
    is_prime = True ##is_prime a variable
    
    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1
        
    if is_prime:
        print("num is a prime number")
    else:
        print("num is not a prime number")
````

**Q6 (Hard).** Write a program that prints the Fibonacci sequence up to `n` terms, where `n` is provided by the user, using a `while` loop.

````python
# Q6
# Q6 (Hard): Print Fibonacci sequence up to n terms using a while loop
n = int(input("Enter the number of terms: "))
n1, n2 = 0, 1
count = 0

if n <= 0:
    print("Please enter a positive integer.")
else:
    while count < n:
        print(n1, end=" ")
        nth = n1 + n2
        n1 = n2
        n2 = nth
        count += 1
````

**Q7 (Hard).** Write a program that checks whether a number is an Armstrong number (e.g., 153 = 1³ + 5³ + 3³) using a `while` loop to extract digits.

````python
# Q7
num = int(input("Enter a number: "))
sum_of_powers = 0
temp = num
power = len(str(num))

while temp > 0:
    digit = temp % 10
    sum_of_powers += digit ** power
    temp //= 10

if num == sum_of_powers:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
````

## Part D: Mini-Challenge

**FizzBuzz, with a twist.** Write a program that loops from 1 to 50 and:
- Prints `"Fizz"` if the number is divisible by 3
- Prints `"Buzz"` if the number is divisible by 5
- Prints `"FizzBuzz"` if divisible by both
- Otherwise prints the number itself
- **Twist:** additionally, keep a running count of how many `"FizzBuzz"` results occurred, and print that count at the end.

This combines `for` loops, `if-elif-else`, the modulus operator, and an accumulator variable.

````python
fizzbuzz_count = 0

for i in range(1, 51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
        fizzbuzz_count += 1
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)

print(f"\nTotal FizzBuzz count: {fizzbuzz_count}")
````
